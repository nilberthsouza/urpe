### Exemplo 3.1

**Problema:**

Pelas tabelas de características elétricas de condutores encordoados de alumínio puro, o condutor **Marigold** de $1.113.000\,\text{CM}$ e de $61$ fios tem:
- Resistência CC a $20\,^{\circ}\text{C}$ de $0,01558 \, \Omega$ por $1.000$ pés;
- Resistência CA a $50\,^{\circ}\text{C}$ de $0,0956 \, \Omega$ por milha.

Verifique a resistência CC e determine a relação entre as resistências CA e CC. Considere:
- Um aumento de $2\%$ por causa do encordoamento.
- O efeito pelicular causando um aumento na resistência.

---

**Solução:**

A equação para a resistência a $20\,^{\circ}\text{C}\$

$R_{20} = \frac{17,0 \times 1000}{1.113 \times 10^3} \times 1,02$


A equação para a resistência a $ 50\,^{\circ}\text{C}: $

$R_{50} = R_{20} \times \frac{228 + 50}{228 + 20}$

A relação entre resistências é dada por:

$
\frac{R}{R_{50}} = \frac{0,0956}{R_{50}}
$

O aumento percentual devido ao efeito pelicular é:
$
\text{Aumento \(= 3,7\%\)}
$

---

### Código em Python

```python

# Definição das variáveis
seccao_transversal_cm = 1113_000  # em CM
resistencia_cc_20 = 0.01558  # em ohms por 1000 pés
resistencia_ca_50_milha = 0.0956  # em ohms por milha
aumento_encordoamento = 1.02  # aumento percentual do encordoamento

# Constantes para cálculo de temperatura
coef_temp_num = 228 + 50
coef_temp_den = 228 + 20
milha_para_pes = 5.280

# Cálculo da resistência a 20°C
r_20 = ((17.0 * 1000) / seccao_transversal_cm) * aumento_encordoamento

# Cálculo da resistência a 50°C
r_50 = resistencia_cc_20 * (coef_temp_num / coef_temp_den)

# Relação entre resistências
relacao_resistencias = resistencia_ca_50_milha / (r_50 * milha_para_pes)


# Exibição dos resultados
print("Resistência CC a 20°C:", round(r_20, 7), "ohms por 1000 pés")
print("Resistência CC a 50°C:", round(r_50, 7), "ohms por 1000 pés")
print("Relação entre resistências (CA/CC):", round(relacao_resistencias, 5))
```
